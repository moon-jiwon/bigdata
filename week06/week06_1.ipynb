## PART 5. 빠진 값(결측치) 처리하기

### 결측치(Missing Value)란?

> 데이터가 **비어있는 경우**를 결측치라고 합니다.  
> `NaN` (Not a Number) 으로 표시됩니다.

### 왜 처리해야 할까요?

- 결측치가 있으면 계산이 틀려질 수 있고 오류가 날 수 있습니다

### 결측치 처리 방법

| 방법 | 코드 | 언제 쓰나 |
|------|------|----------|
| 확인 | `df.isnull().sum()` | 결측치가 몇 개인지 확인 |
| 제거 | `df.dropna()` | 결측치 있는 행 통째로 제거 |
|      | `df.dropna(subset=['컬럼명'])` | 해당 컬럼에 결측치가 있는 행만 제거 |
| 채우기 | `df.fillna(값)` | 결측치를 특정 값으로 채우기 |

````python
df.isnull().sum()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
plt.rc("font", family="Malgun Gothic")
plt.rcParams["axes.unicode_minus"] = False

# 1. 데이터 불러오기 (글자가 깨지면 encoding="cp949")
df = pd.read_csv("서울상권.csv.xz", encoding="utf-8")

# 2. 독서실/스터디카페만 고르기
study = df[df["상권업종소분류명"] == "독서실/스터디 카페"]

# 3. 산점도 그리기
plt.figure(figsize=(10, 8))

# 배경: 서울 전체 상가 (회색으로 연하게 → 서울 모양이 보임)
plt.scatter(df["경도"], df["위도"], s=0.1, c="lightgray", alpha=0.3)

# 독서실/스터디카페 (파란 점)
plt.scatter(study["경도"], study["위도"], s=5, c="blue", alpha=0.5, label="독서실/스터디카페")

# 4. 구 이름 표시 (각 구의 점 평균 위치에)
center = study.groupby("시군구명")[["경도", "위도"]].mean()
for name, row in center.iterrows():
    plt.text(row["경도"], row["위도"], name, fontsize=8, ha="center")

plt.title("서울 독서실/스터디카페 분포")
plt.xlabel("경도")
plt.ylabel("위도")
plt.legend()
plt.show()
````

````python
1. 독서실/스터디카페 고르기
study = df[df["상권업종소분류명"] == "독서실/스터디 카페"]
전체 상가 중 소분류가 독서실/스터디 카페인 것만 남겨둡니다.

2. 가장 많은 구 찾기
gu = study["시군구명"].value_counts()
구별 개수를 많은 순서대로 계산합니다.

3. 산점도로 분포 보기
plt.scatter(study["경도"], study["위도"], ...)
경도와 위도로 scatter를 사용해서 표시합니다.
````

````python
s는 점의 크기고, 숫자가 클수록 점이 커진다는 것을 알게 됐습니다.
alpha는 점의 투명도고, 0이면 투명하고 1이면 진해진다는 것을 알게 됐습니다.
    
center = study.groupby("시군구명")[["경도", "위도"]].mean() => 구마다 독서실/스터디카페 위치의 평균(경도, 위도)을 구해준다는 것을 됐습니다.
for name, row in center.iterrows(): => iterrows()는 표를 한 줄씩 꺼내 주는 함수입니다. 이름도 iterate + rows에서 왔다는 것을 알게 됐습니다.
    plt.text(row["경도"], row["위도"], name, fontsize=8, ha="center") => plt.text(...): ha="center"는 글자의 가운데가 점에 오게 해줍니다.
````

````python
plt.scatter(df["경도"], df["위도"], s=0.1, c="lightgray", alpha=0.3)에서 상가들의 위치를 찍은 거라서 중간에 비어있는 부분이 있었습니다.
그래서 생각을 해보니까 잘못된 게 아니고, 상가가 없는 곳도 있기 때문에 실제 지도와 좀 다르게 생긴 게 맞고, 상가가 있는 곳에만 회색으로 그려져있는 것입니다.
````
